Download the BookCorpus dataset. Take every 7-th sample (the indices are multiple of 7:[0,7,14,21,...]) from the entire dataset.
```python
from datasets import load_dataset
from tokenizers import Tokenizer, models, trainers, pre_tokenizers, normalizers, decoders
import os

# 1. Generator to extract every 7th sample via streaming
def sample_every_nth(dataset_name="bookcorpus/bookcorpus", step=7, batch_size=10000):
    """
    Streams BookCorpus and yields batches of text where the index is a multiple of `step`.
    """
    dataset = load_dataset(dataset_name, split="train", streaming=True,trust_remote_code=True)
    batch = []

    for idx, item in enumerate(dataset):
        if idx % step == 0:
            batch.append(item["text"])
            if len(batch) >= batch_size:
                yield batch
                batch = []

    if batch:
        yield batch
```

```python
from tokenizers.trainers import BpeTrainer
trainer=BpeTrainer(vocab_size=5000,special_tokens=["[PAD]","[UNK]"],continuing_subword_prefix="##")
```

```python
from tokenizers.normalizers import Lowercase
from tokenizers.pre_tokenizers import Whitespace
from tokenizers.models import BPE
from tokenizers import Tokenizer

model=BPE(unk_token="[UNK]")
tokenizer=Tokenizer(model)

tokenizer.normalizer=Lowercase()
tokenizer.pre_tokenizer=Whitespace()
```

```python
tokenizer.train_from_iterator(sample_every_nth(batch_size=10000),trainer=trainer)
```

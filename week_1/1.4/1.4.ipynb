### Translation Dataset

- wmt-14 (workshop on machine translation) is a dataset for machine translation. It contains parallel sentences in multiple languages. The dataset is available in the Hugging Face datasets library and can be loaded using the `load_dataset` function.

- 5 sub-datasets called configurations are available for this dataset. Each configuration contains parallel sentences in a specific language pair. The available configurations are:
  - `cs-en`: Czech to English
  - `de-en`: German to English
  - `fr-en`: French to English
  - `rs-en`: Russian to English
  - `hi-en`: Hindi to English

```python
from datasets import get_dataset_config_names,get_dataset_split_names

print(get_dataset_config_names("wmt/wmt14"))
print(get_dataset_split_names("wmt/wmt14",config_name="hi-en"))
```

- wmt  - is a collection of several datasets
- glue dataset - is a collection of several datasets for natural language understanding tasks. It contains 9 different datasets for various tasks such as 
    - sentiment analysis,
    -  question answering,
    -  and textual entailment.(paraphrasing)
    -  The dataset is available in the Hugging Face datasets library and can be loaded using the `load_dataset` function.

```python
translation_dataset=load_dataset("wmt/wmt14",name="hi-en")
print(translation_dataset)
```

```python
# if we want all the samples as a single dataset(without any splits)

raw_dataset=load_dataset("wmt/wmt14",name="hi-en",split="train+validation+test")
print(raw_dataset)
print(len(raw_dataset)) # 
```

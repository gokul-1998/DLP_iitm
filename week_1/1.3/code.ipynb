# 1. load `stanfordnlp/imbd` dataset

```python
from datasets import load_dataset
imdb_dataset = load_dataset("stanfordnlp/imdb")
print(imdb_dataset)
```

```python
print(imdb_dataset)
```

```python
# how do you get only the train dataset

imdb_train_split=imdb_dataset['train']
print(imdb_train_split)
```

```python
# we want to remove the unsupervised

_ = imdb_dataset.pop('unsupervised')
print(imdb_dataset)
```

```python
# what if we want to download the train split alone?

train_split = load_dataset("stanfordnlp/imdb", split='train')
print(train_split)

```

```python
# do train test split on the train split
small_ds=train_split.train_test_split(test_size=0.2)
print(small_ds)
```

```python
# to store the dataset locally, we can use the `save_to_disk` method

# to store small_ds in  train.csv and test.csv in data folder
small_ds.save_to_disk("data")
```
